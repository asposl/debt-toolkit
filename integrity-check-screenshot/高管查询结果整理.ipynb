```python
import os
import shutil
from pathlib import Path

def reorganize_photos(source_dir, output_dir):
    """
    将 高管名/图片 或 高管名/二级文件夹/图片 的层级结构
    转换为 图片名(或二级文件夹名)/高管名图片(或原文件名) 的层级结构
    """
    src_path = Path(source_dir)
    out_path = Path(output_dir)
    
    # 确保输出总文件夹存在
    out_path.mkdir(parents=True, exist_ok=True)
    
    # 支持的图片格式
    valid_extensions = {'.jpg', '.jpeg', '.png', '.gif', '.bmp', '.webp'}
    
    # 遍历源目录下的每一个高管文件夹
    for exec_folder in src_path.iterdir():
        if not exec_folder.is_dir():
            continue
            
        exec_name = exec_folder.name
        
        # 遍历高管文件夹内的所有子项（可能是文件，也可能是二级文件夹）
        for item in exec_folder.iterdir():
            
            # 情况 1：直接在高管目录下的图片（原逻辑）
            if item.is_file() and item.suffix.lower() in valid_extensions:
                photo_name = item.stem  # 新文件夹名
                photo_ext = item.suffix
                
                new_photo_folder = out_path / photo_name
                new_photo_folder.mkdir(parents=True, exist_ok=True)
                
                # 以高管名重命名
                new_file_name = f"{exec_name}{photo_ext}"
                new_file_path = new_photo_folder / new_file_name
                
                shutil.copy2(item, new_file_path)
                print(f"✅ [一级图片] 已复制: {exec_name}/{item.name} --> {photo_name}/{new_file_name}")
            
            # 情况 2：高管目录下的二级文件夹
            elif item.is_dir():
                sub_folder_name = item.name # 二级文件夹名称（作为新的目标文件夹名）
                
                # 遍历二级文件夹内部的图片
                for sub_item in item.iterdir():
                    if sub_item.is_file() and sub_item.suffix.lower() in valid_extensions:
                        
                        new_photo_folder = out_path / sub_folder_name
                        new_photo_folder.mkdir(parents=True, exist_ok=True)
                        
                        # 根据你的需求：由于子文件名称本来就有区别，直接用原文件名，不用担心覆盖
                        new_file_name = sub_item.name 
                        new_file_path = new_photo_folder / new_file_name
                        
                        shutil.copy2(sub_item, new_file_path)
                        print(f"📂 [二级目录] 已归类: {exec_name}/{sub_folder_name}/{sub_item.name} --> {sub_folder_name}/{new_file_name}")

if __name__ == "__main__":
    # ------------------ 设置区 ------------------
    SOURCE_DIRECTORY = r"项目目录\诚信截图\截图\dist\IBDW\djg"
    OUTPUT_DIRECTORY = r"项目目录\诚信截图\截图\dist\IBDW"
    # --------------------------------------------
    
    print("开始处理文件...\n")
    reorganize_photos(SOURCE_DIRECTORY, OUTPUT_DIRECTORY)
    print("\n🎉 处理完成！请前往输出文件夹检查结果。")
```
